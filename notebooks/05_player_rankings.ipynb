# 05. Player Rankings & PRI Scoring
Calculates final PRI-Bat and PRI-Bowl scores and exports Top 10 rankings.

```python
import pandas as pd
import sys
sys.path.append('../src')
from data_loader import load_raw_data
from preprocessing import preprocess_data
from pressure_detector import detect_pressure_situations
from batting_analysis import analyze_batting_performance
from bowling_analysis import analyze_bowling_performance
from scoring_engine import score_batting_pri, score_bowling_pri

data = load_raw_data('../data/raw')
bbb, _ = preprocess_data(data)
bbb_p = detect_pressure_situations(bbb)
bat = analyze_batting_performance(bbb_p)
bowl = analyze_bowling_performance(bbb_p)

top_bat = score_batting_pri(bat).head(10)
top_bowl = score_bowling_pri(bowl).head(10)

print('Top 10 Batters:')
print(top_bat[['rank', 'player_name', 'pri_score', 'p_strike_rate', 'delta_strike_rate']])
print('
Top 10 Bowlers:')
print(top_bowl[['rank', 'player_name', 'pri_score', 'p_economy', 'delta_economy']])
```
