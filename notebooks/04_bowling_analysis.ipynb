# 04. Bowling Performance Under Pressure
Aggregates pressure vs normal baseline metrics for bowlers.

```python
import pandas as pd
import sys
sys.path.append('../src')
from data_loader import load_raw_data
from preprocessing import preprocess_data
from pressure_detector import detect_pressure_situations
from bowling_analysis import analyze_bowling_performance

data = load_raw_data('../data/raw')
bbb, _ = preprocess_data(data)
bbb_p = detect_pressure_situations(bbb)
bowl_profile = analyze_bowling_performance(bbb_p)

print('Top bowlers by pressure legal balls bowled:')
print(bowl_profile.sort_values(by='p_balls', ascending=False)[['player_name', 'p_balls', 'p_matches', 'p_economy', 'delta_economy', 'p_wickets']].head(10))
```
