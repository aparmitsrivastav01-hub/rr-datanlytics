# 03. Batting Performance Under Pressure
Aggregates pressure vs normal baseline metrics for batters.

```python
import pandas as pd
import sys
sys.path.append('../src')
from data_loader import load_raw_data
from preprocessing import preprocess_data
from pressure_detector import detect_pressure_situations
from batting_analysis import analyze_batting_performance

data = load_raw_data('../data/raw')
bbb, _ = preprocess_data(data)
bbb_p = detect_pressure_situations(bbb)
bat_profile = analyze_batting_performance(bbb_p)

print('Top batters by pressure balls faced:')
print(bat_profile.sort_values(by='p_balls', ascending=False)[['player_name', 'p_balls', 'p_matches', 'p_strike_rate', 'delta_strike_rate']].head(10))
```
